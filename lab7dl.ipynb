```python
# =====================================================================
# CS3807 - Deep Learning Lab | Experiment 7
# FC-AE, CAE, Denoising CAE, VAE on MNIST  (single-cell Kaggle script)
# Kaggle: turn on GPU (Settings > Accelerator), paste into ONE cell, Run.
# All figures are shown inline AND saved to /kaggle/working/exp7_outputs/
# =====================================================================
import os, time, warnings
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from skimage.metrics import structural_similarity as ssim_fn

warnings.filterwarnings("ignore")
keras.utils.set_random_seed(42)
rng = np.random.default_rng(42)

OUT = "/kaggle/working/exp7_outputs" if os.path.exists("/kaggle/working") else "exp7_outputs"
os.makedirs(OUT, exist_ok=True)

# ----------------------------- Settings ------------------------------
N_TRAIN, N_TEST = 10000, 2000      # set N_TRAIN=60000, N_TEST=10000 for full MNIST
EPOCHS, BATCH = 20, 128
VAE_EPOCHS = 30
VAL_SPLIT = 0.1                    # validation is carved out of TRAIN; test stays unseen
LATENT_DIMS = [2, 8, 16, 32]
GAUSS_SIGMAS = [0.1, 0.2, 0.3]
SP_PROBS = [0.05, 0.10, 0.20]
TRAIN_SIGMA, TRAIN_SP = 0.2, 0.10  # noise level used to TRAIN the denoisers

# ------------------------------- Data --------------------------------
(x_tr, y_tr), (x_te, y_te) = keras.datasets.mnist.load_data()
x_tr = (x_tr[:N_TRAIN].astype("float32") / 255.)[..., None]
x_te = (x_te[:N_TEST].astype("float32") / 255.)[..., None]
y_te = y_te[:N_TEST]
x_tr_flat, x_te_flat = x_tr.reshape(len(x_tr), -1), x_te.reshape(len(x_te), -1)
print("Train:", x_tr.shape, "Test:", x_te.shape, "| range:", x_tr.min(), x_tr.max())

# ------------------------------ Helpers ------------------------------
def save_show(name):
    plt.savefig(f"{OUT}/{name}.png", dpi=150, bbox_inches="tight")
    plt.show()

def evaluate(x, xh):
    """MSE, MAE, mean SSIM over a set of images shaped (N,28,28,1)."""
    x, xh = x.reshape(-1, 28, 28), xh.reshape(-1, 28, 28)
    mse = float(np.mean((x - xh) ** 2))
    mae = float(np.mean(np.abs(x - xh)))
    ss = float(np.mean([ssim_fn(a, b, data_range=1.0) for a, b in zip(x, xh)]))
    return dict(MSE=mse, MAE=mae, SSIM=ss)

def per_image_error(x, xh):
    return np.mean((x.reshape(len(x), -1) - xh.reshape(len(xh), -1)) ** 2, axis=1)

def image_grid(rows, row_titles, n=10, cmaps=None, title=None, fname=None):
    fig, ax = plt.subplots(len(rows), n, figsize=(1.3 * n, 1.5 * len(rows)))
    ax = np.atleast_2d(ax)
    for r, (imgs, rt) in enumerate(zip(rows, row_titles)):
        for c in range(n):
            ax[r, c].imshow(imgs[c].reshape(28, 28), cmap=(cmaps[r] if cmaps else "gray"),
                            vmin=0, vmax=1)
            ax[r, c].axis("off")
        ax[r, 0].set_title(rt, fontsize=9, loc="left")
    if title: fig.suptitle(title, y=1.0)
    plt.tight_layout()
    if fname: save_show(fname)
    else: plt.show()

def plot_loss(hist, title, fname, keys=("loss", "val_loss"), ylabel="Reconstruction loss (BCE)"):
    plt.figure(figsize=(6, 4))
    for k, lab in zip(keys, ["Training loss", "Validation loss"]):
        plt.plot(hist.history[k], label=lab)
    plt.xlabel("Epoch"); plt.ylabel(ylabel); plt.title(title); plt.legend(); plt.grid(alpha=.3)
    save_show(fname)

def fit_timed(model, x, y, epochs=EPOCHS):
    t = time.time()
    h = model.fit(x, y, epochs=epochs, batch_size=BATCH, validation_split=VAL_SPLIT,
                  shuffle=True, verbose=2)
    return h, time.time() - t

results = {}  # consolidated results

# ============== 1. FULLY CONNECTED AUTOENCODER (+ latent study) ==============
def build_fc(dz):
    inp = keras.Input((784,))
    h = layers.Dense(128, activation="relu")(inp)
    h = layers.Dense(32, activation="relu")(h)
    z = layers.Dense(dz, activation="relu", name="latent")(h)   # spec: ReLU hidden layers
    h = layers.Dense(32, activation="relu")(z)
    h = layers.Dense(128, activation="relu")(h)
    out = layers.Dense(784, activation="sigmoid")(h)
    m = keras.Model(inp, out, name=f"FC_AE_z{dz}")
    m.compile(optimizer=keras.optimizers.Adam(1e-3), loss="binary_crossentropy")
    return m

latent_rows, fc_hist, fc_ae = [], None, None
for dz in LATENT_DIMS:
    print(f"\n=== FC Autoencoder, latent dim = {dz} ===")
    m = build_fc(dz)
    h, t = fit_timed(m, x_tr_flat, x_tr_flat)
    rec = m.predict(x_te_flat, verbose=0)
    met = evaluate(x_te, rec)
    latent_rows.append(dict(LatentDim=dz, MSE=met["MSE"], SSIM=met["SSIM"], Parameters=m.count_params()))
    if dz == 16:
        fc_ae, fc_hist, fc_rec, fc_time, fc_met = m, h, rec, t, met
        results["FC Autoencoder"] = dict(**met, Parameters=m.count_params(), Time_s=t)

# Plot 1: original vs reconstruction vs error
rec_img = fc_rec.reshape(-1, 28, 28, 1)
err_img = np.abs(x_te - rec_img)
image_grid([x_te[:10], rec_img[:10], err_img[:10]],
           ["Original", "FC-AE reconstruction", "Absolute error"],
           cmaps=["gray", "gray", "hot"], title="Plot 1: Original vs Reconstructed (FC-AE, z=16)",
           fname="plot1_fc_ae_reconstruction")
# Plot 2: training / validation loss
plot_loss(fc_hist, "Plot 2: FC-AE training vs validation loss", "plot2_fc_ae_loss")
print("Final train loss: %.4f | final val loss: %.4f" %
      (fc_hist.history["loss"][-1], fc_hist.history["val_loss"][-1]))
print("FC-AE test metrics:", fc_met)

# ===================== 2. CONVOLUTIONAL AUTOENCODER =====================
def build_cae(name="CAE"):
    inp = keras.Input((28, 28, 1))
    x = layers.Conv2D(32, 3, activation="relu", padding="same")(inp)
    x = layers.MaxPooling2D(2, padding="same")(x)
    x = layers.Conv2D(64, 3, activation="relu", padding="same")(x)
    x = layers.MaxPooling2D(2, padding="same")(x)                 # 7x7x64 latent map
    x = layers.Conv2D(64, 3, activation="relu", padding="same")(x)
    x = layers.UpSampling2D(2)(x)
    x = layers.Conv2D(32, 3, activation="relu", padding="same")(x)
    x = layers.UpSampling2D(2)(x)
    out = layers.Conv2D(1, 3, activation="sigmoid", padding="same")(x)
    m = keras.Model(inp, out, name=name)
    m.compile(optimizer=keras.optimizers.Adam(1e-3), loss="binary_crossentropy")
    return m

print("\n=== Convolutional Autoencoder ===")
cae = build_cae("CAE")
cae_hist, cae_time = fit_timed(cae, x_tr, x_tr)
cae_rec = cae.predict(x_te, verbose=0)
cae_met = evaluate(x_te, cae_rec)
results["Conv. Autoencoder"] = dict(**cae_met, Parameters=cae.count_params(), Time_s=cae_time)

cmp = pd.DataFrame([
    dict(Model="Fully Connected AE", **fc_met, Parameters=fc_ae.count_params()),
    dict(Model="Convolutional AE", **cae_met, Parameters=cae.count_params())])
print("\nFC-AE vs CAE:\n", cmp.to_string(index=False))

# Plot 3: Original -> FC-AE -> CAE
image_grid([x_te[:10], rec_img[:10], cae_rec[:10]],
           ["Original", "FC-AE", "CAE"],
           title="Plot 3: Reconstruction comparison", fname="plot3_fc_vs_cae")
plot_loss(cae_hist, "CAE training vs validation loss", "cae_loss")

# ===================== 3. DENOISING CONVOLUTIONAL AE =====================
def add_gaussian(x, sigma, seed=0):
    r = np.random.default_rng(seed)
    return np.clip(x + r.normal(0, sigma, x.shape), 0, 1).astype("float32")

def add_salt_pepper(x, p, seed=0):
    r = np.random.default_rng(seed)
    u = r.random(x.shape)
    out = x.copy()
    out[u < p / 2] = 0.0                  # pepper
    out[(u >= p / 2) & (u < p)] = 1.0     # salt
    return out.astype("float32")

print("\n=== Denoising CAE (Gaussian, sigma=%.1f) ===" % TRAIN_SIGMA)
dcae_g = build_cae("DCAE_gaussian")
h_g, t_g = fit_timed(dcae_g, add_gaussian(x_tr, TRAIN_SIGMA, 1), x_tr)   # target = CLEAN image
print("\n=== Denoising CAE (Salt & Pepper, p=%.2f) ===" % TRAIN_SP)
dcae_s = build_cae("DCAE_saltpepper")
h_s, t_s = fit_timed(dcae_s, add_salt_pepper(x_tr, TRAIN_SP, 2), x_tr)

# Plot 4: clean | noisy | denoised for both noise types
xn_g = add_gaussian(x_te, TRAIN_SIGMA, 10); xd_g = dcae_g.predict(xn_g, verbose=0)
xn_s = add_salt_pepper(x_te, TRAIN_SP, 11); xd_s = dcae_s.predict(xn_s, verbose=0)
image_grid([x_te[:10], xn_g[:10], xd_g[:10]], ["Clean", f"Noisy (Gaussian σ={TRAIN_SIGMA})", "Denoised"],
           title="Plot 4a: Clean vs Noisy vs Denoised (Gaussian)", fname="plot4a_denoise_gaussian")
image_grid([x_te[:10], xn_s[:10], xd_s[:10]], ["Clean", f"Noisy (Salt&Pepper p={TRAIN_SP})", "Denoised"],
           title="Plot 4b: Clean vs Noisy vs Denoised (Salt & Pepper)", fname="plot4b_denoise_saltpepper")

# Plot 5: noise level vs quality
noise_rows = []
for s in GAUSS_SIGMAS:
    xn = add_gaussian(x_te, s, 100 + int(s * 10)); xd = dcae_g.predict(xn, verbose=0)
    noise_rows.append(dict(Type="Gaussian", Level=s,
        **{f"Noisy_{k}": v for k, v in evaluate(x_te, xn).items()},
        **{f"Denoised_{k}": v for k, v in evaluate(x_te, xd).items()}))
for p in SP_PROBS:
    xn = add_salt_pepper(x_te, p, 200 + int(p * 100)); xd = dcae_s.predict(xn, verbose=0)
    noise_rows.append(dict(Type="Salt&Pepper", Level=p,
        **{f"Noisy_{k}": v for k, v in evaluate(x_te, xn).items()},
        **{f"Denoised_{k}": v for k, v in evaluate(x_te, xd).items()}))
noise_df = pd.DataFrame(noise_rows)
print("\nNoise level vs reconstruction quality (vs clean image):\n", noise_df.round(4).to_string(index=False))

for ntype in ["Gaussian", "Salt&Pepper"]:
    d = noise_df[noise_df.Type == ntype]
    fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
    for a, metric in zip(ax, ["MSE", "MAE", "SSIM"]):          # separate axes per metric
        a.plot(d.Level, d[f"Noisy_{metric}"], "o--", label="Noisy input")
        a.plot(d.Level, d[f"Denoised_{metric}"], "s-", label="Denoised")
        a.set_xlabel("σ" if ntype == "Gaussian" else "corruption p"); a.set_ylabel(metric)
        a.set_title(metric); a.grid(alpha=.3); a.legend()
    fig.suptitle(f"Plot 5: Noise level vs quality ({ntype})")
    plt.tight_layout(); save_show(f"plot5_noise_vs_quality_{ntype.replace('&','')}")

# Denoising CAE row for consolidated table (Gaussian sigma=0.2 test set vs clean)
dcae_met = evaluate(x_te, xd_g)
results["Denoising CAE"] = dict(**dcae_met, Parameters=dcae_g.count_params(), Time_s=t_g)
print("Denoising CAE (Gaussian σ=0.2) test metrics:", dcae_met)

# =================== 4. VARIATIONAL AUTOENCODER (z in R^2) ===================
class Sampling(layers.Layer):
    """Reparameterization trick: z = mu + sigma * eps, eps ~ N(0, I)."""
    def call(self, inputs):
        mu, logvar = inputs
        eps = tf.random.normal(tf.shape(mu))
        return mu + tf.exp(0.5 * logvar) * eps

def build_encoder(dz=2):
    inp = keras.Input((28, 28, 1))
    x = layers.Conv2D(32, 3, strides=2, padding="same", activation="relu")(inp)   # 14x14
    x = layers.Conv2D(64, 3, strides=2, padding="same", activation="relu")(x)     # 7x7
    x = layers.Flatten()(x)
    x = layers.Dense(16, activation="relu")(x)
    mu = layers.Dense(dz, name="mu")(x)
    logvar = layers.Dense(dz, name="log_var")(x)
    z = Sampling()([mu, logvar])
    return keras.Model(inp, [mu, logvar, z], name="encoder")

def build_decoder(dz=2):
    inp = keras.Input((dz,))
    x = layers.Dense(7 * 7 * 64, activation="relu")(inp)
    x = layers.Reshape((7, 7, 64))(x)
    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same", activation="relu")(x)
    x = layers.Conv2DTranspose(32, 3, strides=2, padding="same", activation="relu")(x)
    out = layers.Conv2DTranspose(1, 3, padding="same", activation="sigmoid")(x)
    return keras.Model(inp, out, name="decoder")

class VAE(keras.Model):
    def __init__(self, encoder, decoder, beta=1.0, **kw):
        super().__init__(**kw)
        self.encoder, self.decoder, self.beta = encoder, decoder, beta
        self.t_total = keras.metrics.Mean(name="loss")
        self.t_rec = keras.metrics.Mean(name="rec_loss")
        self.t_kl = keras.metrics.Mean(name="kl_loss")

    @property
    def metrics(self):
        return [self.t_total, self.t_rec, self.t_kl]

    def call(self, x):
        return self.decoder(self.encoder(x)[2])

    def _losses(self, x):
        mu, logvar, z = self.encoder(x)
        xh = self.decoder(z)
        bce = keras.losses.binary_crossentropy(x, xh)                 # (B,28,28)
        rec = tf.reduce_mean(tf.reduce_sum(bce, axis=(1, 2)))         # sum over pixels, mean over batch
        kl = -0.5 * tf.reduce_mean(tf.reduce_sum(1 + logvar - tf.square(mu) - tf.exp(logvar), axis=1))
        return rec, kl

    def _update(self, rec, kl):
        total = rec + self.beta * kl
        self.t_total.update_state(total); self.t_rec.update_state(rec); self.t_kl.update_state(kl)
        return total

    def train_step(self, data):
        x = data[0] if isinstance(data, (tuple, list)) else data
        with tf.GradientTape() as tape:
            rec, kl = self._losses(x)
            total = rec + self.beta * kl
        grads = tape.gradient(total, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))
        self._update(rec, kl)
        return {m.name: m.result() for m in self.metrics}

    def test_step(self, data):
        x = data[0] if isinstance(data, (tuple, list)) else data
        rec, kl = self._losses(x)
        self._update(rec, kl)
        return {m.name: m.result() for m in self.metrics}

print("\n=== Variational Autoencoder (latent dim = 2) ===")
enc, dec = build_encoder(2), build_decoder(2)
vae = VAE(enc, dec, beta=1.0, name="VAE")
vae.compile(optimizer=keras.optimizers.Adam(1e-3))
t0 = time.time()
vae_hist = vae.fit(x_tr, x_tr, epochs=VAE_EPOCHS, batch_size=BATCH,
                   validation_split=VAL_SPLIT, shuffle=True, verbose=2)
vae_time = time.time() - t0
vae_params = enc.count_params() + dec.count_params()

# Reconstructions use the posterior mean (z = mu)
mu_te, lv_te, _ = enc.predict(x_te, verbose=0)
vae_rec = dec.predict(mu_te, verbose=0)
vae_met = evaluate(x_te, vae_rec)
results["VAE"] = dict(**vae_met, Parameters=vae_params, Time_s=vae_time)

# VAE test losses (per image, summed over pixels)
eps_ = 1e-7
xt = x_te.reshape(len(x_te), -1); xr = np.clip(vae_rec.reshape(len(x_te), -1), eps_, 1 - eps_)
vae_rec_loss = float(np.mean(-np.sum(xt * np.log(xr) + (1 - xt) * np.log(1 - xr), axis=1)))
vae_kl = float(np.mean(-0.5 * np.sum(1 + lv_te - mu_te ** 2 - np.exp(lv_te), axis=1)))
vae_table = pd.DataFrame({"Metric": ["Reconstruction Loss", "KL Loss", "Total Loss", "Test MSE", "Test MAE", "Mean SSIM"],
                          "Value": [vae_rec_loss, vae_kl, vae_rec_loss + vae_kl,
                                    vae_met["MSE"], vae_met["MAE"], vae_met["SSIM"]]})
print("\nVAE metrics:\n", vae_table.round(4).to_string(index=False))

# Plot: VAE training / validation loss
fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
for a, (k, t) in zip(ax, [("loss", "Total loss"), ("rec_loss", "Reconstruction loss"), ("kl_loss", "KL loss")]):
    a.plot(vae_hist.history[k], label="Training"); a.plot(vae_hist.history["val_" + k], label="Validation")
    a.set_title(t); a.set_xlabel("Epoch"); a.set_ylabel(t); a.legend(); a.grid(alpha=.3)
fig.suptitle("VAE training / validation loss"); plt.tight_layout(); save_show("vae_loss_curves")

# Plot 6: 2-D latent space
plt.figure(figsize=(7, 6))
sc = plt.scatter(mu_te[:, 0], mu_te[:, 1], c=y_te, cmap="tab10", s=8, alpha=.8)
plt.colorbar(sc, ticks=range(10), label="Digit label")
plt.xlabel("z1"); plt.ylabel("z2"); plt.title("Plot 6: VAE latent space (test set, μ)"); plt.grid(alpha=.3)
save_show("plot6_vae_latent_space")

# Plot 7: 25 random samples z ~ N(0, I)
z_rand = np.random.default_rng(7).standard_normal((25, 2)).astype("float32")
gen = dec.predict(z_rand, verbose=0)
fig, ax = plt.subplots(5, 5, figsize=(6, 6))
for i, a in enumerate(ax.ravel()):
    a.imshow(gen[i].reshape(28, 28), cmap="gray", vmin=0, vmax=1); a.axis("off")
fig.suptitle("Plot 7: Randomly generated VAE images (z ~ N(0,I))"); plt.tight_layout()
save_show("plot7_vae_generated")

# Plot 8: latent interpolation (between encoded test digits) + a random pair
alphas = np.linspace(0, 1, 11)
def interp_row(zA, zB):
    zs = np.array([(1 - a) * zA + a * zB for a in alphas], dtype="float32")
    return dec.predict(zs, verbose=0)
pairs = [(3, 8), (1, 7)]
rows, titles = [], []
for a_, b_ in pairs:
    zA, zB = mu_te[np.where(y_te == a_)[0][0]], mu_te[np.where(y_te == b_)[0][0]]
    rows.append(interp_row(zA, zB)); titles.append(f"Digit {a_} → {b_}")
rows.append(interp_row(z_rand[0], z_rand[1])); titles.append("Random pair")
image_grid(rows, titles, n=11, title="Plot 8: Latent-space interpolation (α = 0 → 1)", fname="plot8_vae_interpolation")

# =============== Reconstruction error distribution & high-error ===============
def error_analysis(name, xh, tag):
    e = per_image_error(x_te, xh)
    plt.figure(figsize=(6, 4))
    plt.hist(e, bins=50, color="steelblue", edgecolor="k", alpha=.85)
    plt.axvline(e.mean(), color="r", ls="--", label=f"mean={e.mean():.4f}")
    plt.axvline(np.median(e), color="g", ls="--", label=f"median={np.median(e):.4f}")
    plt.xlabel("Per-image MSE $e_i$"); plt.ylabel("Number of test images")
    plt.title(f"Plot 9: Reconstruction-error distribution ({name})"); plt.legend()
    save_show(f"plot9_error_distribution_{tag}")
    print(f"[{name}] mean={e.mean():.4f} median={np.median(e):.4f} std={e.std():.4f} "
          f"5-95%=({np.percentile(e,5):.4f},{np.percentile(e,95):.4f}) max={e.max():.4f}")
    top = np.argsort(e)[-5:][::-1]
    image_grid([x_te[top], xh[top]], ["Original", "Reconstruction"], n=5,
               title=f"High-error samples ({name}) | e = " + ", ".join(f"{e[i]:.3f}" for i in top) +
                     " | labels = " + ",".join(str(y_te[i]) for i in top),
               fname=f"high_error_{tag}")

error_analysis("FC-AE", rec_img, "fc_ae")
error_analysis("CAE", cae_rec, "cae")
error_analysis("VAE", vae_rec, "vae")

# ====================== Latent-dimension study (Plot 10) ======================
lat_df = pd.DataFrame(latent_rows)
print("\nLatent dimension study (FC-AE):\n", lat_df.round(5).to_string(index=False))
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(lat_df.LatentDim, lat_df.MSE, "o-"); ax[0].set_xscale("log", base=2)
ax[0].set_xlabel("Latent dimension"); ax[0].set_ylabel("Test MSE"); ax[0].set_title("Latent dim vs MSE"); ax[0].grid(alpha=.3)
ax[1].plot(lat_df.LatentDim, lat_df.SSIM, "o-", color="darkorange"); ax[1].set_xscale("log", base=2)
ax[1].set_xlabel("Latent dimension"); ax[1].set_ylabel("Mean SSIM"); ax[1].set_title("Latent dim vs SSIM"); ax[1].grid(alpha=.3)
fig.suptitle("Plot 10: Effect of latent dimension (FC-AE)"); plt.tight_layout(); save_show("plot10_latent_dim_study")

# =========================== Consolidated tables ===========================
cons = pd.DataFrame(results).T[["MSE", "MAE", "SSIM", "Parameters", "Time_s"]]
cons.index.name = "Model"
print("\n================ CONSOLIDATED RESULTS ================")
print(cons.round(5).to_string())
print("\n(Denoising CAE row = Gaussian σ=0.2 noisy test input vs clean target; VAE row = reconstruction with z=μ)")

cons.to_csv(f"{OUT}/consolidated_results.csv")
lat_df.to_csv(f"{OUT}/latent_dim_study.csv", index=False)
noise_df.to_csv(f"{OUT}/noise_level_results.csv", index=False)
cmp.to_csv(f"{OUT}/fc_vs_cae.csv", index=False)
vae_table.to_csv(f"{OUT}/vae_metrics.csv", index=False)
print(f"\nAll figures and CSV tables saved in: {OUT}")
```

```python
# Continuation: run ONLY if the earlier cell already finished the FC-AE/CAE/denoising parts in this same kernel session.
# =================== 4. VARIATIONAL AUTOENCODER (z in R^2) ===================
class Sampling(layers.Layer):
    """Reparameterization trick: z = mu + sigma * eps, eps ~ N(0, I)."""
    def call(self, inputs):
        mu, logvar = inputs
        eps = tf.random.normal(tf.shape(mu))
        return mu + tf.exp(0.5 * logvar) * eps

def build_encoder(dz=2):
    inp = keras.Input((28, 28, 1))
    x = layers.Conv2D(32, 3, strides=2, padding="same", activation="relu")(inp)   # 14x14
    x = layers.Conv2D(64, 3, strides=2, padding="same", activation="relu")(x)     # 7x7
    x = layers.Flatten()(x)
    x = layers.Dense(16, activation="relu")(x)
    mu = layers.Dense(dz, name="mu")(x)
    logvar = layers.Dense(dz, name="log_var")(x)
    z = Sampling()([mu, logvar])
    return keras.Model(inp, [mu, logvar, z], name="encoder")

def build_decoder(dz=2):
    inp = keras.Input((dz,))
    x = layers.Dense(7 * 7 * 64, activation="relu")(inp)
    x = layers.Reshape((7, 7, 64))(x)
    x = layers.Conv2DTranspose(64, 3, strides=2, padding="same", activation="relu")(x)
    x = layers.Conv2DTranspose(32, 3, strides=2, padding="same", activation="relu")(x)
    out = layers.Conv2DTranspose(1, 3, padding="same", activation="sigmoid")(x)
    return keras.Model(inp, out, name="decoder")

class VAE(keras.Model):
    def __init__(self, encoder, decoder, beta=1.0, **kw):
        super().__init__(**kw)
        self.encoder, self.decoder, self.beta = encoder, decoder, beta
        self.t_total = keras.metrics.Mean(name="loss")
        self.t_rec = keras.metrics.Mean(name="rec_loss")
        self.t_kl = keras.metrics.Mean(name="kl_loss")

    @property
    def metrics(self):
        return [self.t_total, self.t_rec, self.t_kl]

    def call(self, x):
        return self.decoder(self.encoder(x)[2])

    def _vae_losses(self, x):
        mu, logvar, z = self.encoder(x)
        xh = self.decoder(z)
        bce = keras.losses.binary_crossentropy(x, xh)                 # (B,28,28)
        rec = tf.reduce_mean(tf.reduce_sum(bce, axis=(1, 2)))         # sum over pixels, mean over batch
        kl = -0.5 * tf.reduce_mean(tf.reduce_sum(1 + logvar - tf.square(mu) - tf.exp(logvar), axis=1))
        return rec, kl

    def _vae_update(self, rec, kl):
        total = rec + self.beta * kl
        self.t_total.update_state(total); self.t_rec.update_state(rec); self.t_kl.update_state(kl)
        return total

    def train_step(self, data):
        x = data[0] if isinstance(data, (tuple, list)) else data
        with tf.GradientTape() as tape:
            rec, kl = self._vae_losses(x)
            total = rec + self.beta * kl
        grads = tape.gradient(total, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))
        self._vae_update(rec, kl)
        return {m.name: m.result() for m in self.metrics}

    def test_step(self, data):
        x = data[0] if isinstance(data, (tuple, list)) else data
        rec, kl = self._vae_losses(x)
        self._vae_update(rec, kl)
        return {m.name: m.result() for m in self.metrics}

print("\n=== Variational Autoencoder (latent dim = 2) ===")
enc, dec = build_encoder(2), build_decoder(2)
vae = VAE(enc, dec, beta=1.0, name="VAE")
vae.compile(optimizer=keras.optimizers.Adam(1e-3))
t0 = time.time()
vae_hist = vae.fit(x_tr, x_tr, epochs=VAE_EPOCHS, batch_size=BATCH,
                   validation_split=VAL_SPLIT, shuffle=True, verbose=2)
vae_time = time.time() - t0
vae_params = enc.count_params() + dec.count_params()

# Reconstructions use the posterior mean (z = mu)
mu_te, lv_te, _ = enc.predict(x_te, verbose=0)
vae_rec = dec.predict(mu_te, verbose=0)
vae_met = evaluate(x_te, vae_rec)
results["VAE"] = dict(**vae_met, Parameters=vae_params, Time_s=vae_time)

# VAE test losses (per image, summed over pixels)
eps_ = 1e-7
xt = x_te.reshape(len(x_te), -1); xr = np.clip(vae_rec.reshape(len(x_te), -1), eps_, 1 - eps_)
vae_rec_loss = float(np.mean(-np.sum(xt * np.log(xr) + (1 - xt) * np.log(1 - xr), axis=1)))
vae_kl = float(np.mean(-0.5 * np.sum(1 + lv_te - mu_te ** 2 - np.exp(lv_te), axis=1)))
vae_table = pd.DataFrame({"Metric": ["Reconstruction Loss", "KL Loss", "Total Loss", "Test MSE", "Test MAE", "Mean SSIM"],
                          "Value": [vae_rec_loss, vae_kl, vae_rec_loss + vae_kl,
                                    vae_met["MSE"], vae_met["MAE"], vae_met["SSIM"]]})
print("\nVAE metrics:\n", vae_table.round(4).to_string(index=False))

# Plot: VAE training / validation loss
fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
for a, (k, t) in zip(ax, [("loss", "Total loss"), ("rec_loss", "Reconstruction loss"), ("kl_loss", "KL loss")]):
    a.plot(vae_hist.history[k], label="Training"); a.plot(vae_hist.history["val_" + k], label="Validation")
    a.set_title(t); a.set_xlabel("Epoch"); a.set_ylabel(t); a.legend(); a.grid(alpha=.3)
fig.suptitle("VAE training / validation loss"); plt.tight_layout(); save_show("vae_loss_curves")

# Plot 6: 2-D latent space
plt.figure(figsize=(7, 6))
sc = plt.scatter(mu_te[:, 0], mu_te[:, 1], c=y_te, cmap="tab10", s=8, alpha=.8)
plt.colorbar(sc, ticks=range(10), label="Digit label")
plt.xlabel("z1"); plt.ylabel("z2"); plt.title("Plot 6: VAE latent space (test set, μ)"); plt.grid(alpha=.3)
save_show("plot6_vae_latent_space")

# Plot 7: 25 random samples z ~ N(0, I)
z_rand = np.random.default_rng(7).standard_normal((25, 2)).astype("float32")
gen = dec.predict(z_rand, verbose=0)
fig, ax = plt.subplots(5, 5, figsize=(6, 6))
for i, a in enumerate(ax.ravel()):
    a.imshow(gen[i].reshape(28, 28), cmap="gray", vmin=0, vmax=1); a.axis("off")
fig.suptitle("Plot 7: Randomly generated VAE images (z ~ N(0,I))"); plt.tight_layout()
save_show("plot7_vae_generated")

# Plot 8: latent interpolation (between encoded test digits) + a random pair
alphas = np.linspace(0, 1, 11)
def interp_row(zA, zB):
    zs = np.array([(1 - a) * zA + a * zB for a in alphas], dtype="float32")
    return dec.predict(zs, verbose=0)
pairs = [(3, 8), (1, 7)]
rows, titles = [], []
for a_, b_ in pairs:
    zA, zB = mu_te[np.where(y_te == a_)[0][0]], mu_te[np.where(y_te == b_)[0][0]]
    rows.append(interp_row(zA, zB)); titles.append(f"Digit {a_} → {b_}")
rows.append(interp_row(z_rand[0], z_rand[1])); titles.append("Random pair")
image_grid(rows, titles, n=11, title="Plot 8: Latent-space interpolation (α = 0 → 1)", fname="plot8_vae_interpolation")

# =============== Reconstruction error distribution & high-error ===============
def error_analysis(name, xh, tag):
    e = per_image_error(x_te, xh)
    plt.figure(figsize=(6, 4))
    plt.hist(e, bins=50, color="steelblue", edgecolor="k", alpha=.85)
    plt.axvline(e.mean(), color="r", ls="--", label=f"mean={e.mean():.4f}")
    plt.axvline(np.median(e), color="g", ls="--", label=f"median={np.median(e):.4f}")
    plt.xlabel("Per-image MSE $e_i$"); plt.ylabel("Number of test images")
    plt.title(f"Plot 9: Reconstruction-error distribution ({name})"); plt.legend()
    save_show(f"plot9_error_distribution_{tag}")
    print(f"[{name}] mean={e.mean():.4f} median={np.median(e):.4f} std={e.std():.4f} "
          f"5-95%=({np.percentile(e,5):.4f},{np.percentile(e,95):.4f}) max={e.max():.4f}")
    top = np.argsort(e)[-5:][::-1]
    image_grid([x_te[top], xh[top]], ["Original", "Reconstruction"], n=5,
               title=f"High-error samples ({name}) | e = " + ", ".join(f"{e[i]:.3f}" for i in top) +
                     " | labels = " + ",".join(str(y_te[i]) for i in top),
               fname=f"high_error_{tag}")

error_analysis("FC-AE", rec_img, "fc_ae")
error_analysis("CAE", cae_rec, "cae")
error_analysis("VAE", vae_rec, "vae")

# ====================== Latent-dimension study (Plot 10) ======================
lat_df = pd.DataFrame(latent_rows)
print("\nLatent dimension study (FC-AE):\n", lat_df.round(5).to_string(index=False))
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(lat_df.LatentDim, lat_df.MSE, "o-"); ax[0].set_xscale("log", base=2)
ax[0].set_xlabel("Latent dimension"); ax[0].set_ylabel("Test MSE"); ax[0].set_title("Latent dim vs MSE"); ax[0].grid(alpha=.3)
ax[1].plot(lat_df.LatentDim, lat_df.SSIM, "o-", color="darkorange"); ax[1].set_xscale("log", base=2)
ax[1].set_xlabel("Latent dimension"); ax[1].set_ylabel("Mean SSIM"); ax[1].set_title("Latent dim vs SSIM"); ax[1].grid(alpha=.3)
fig.suptitle("Plot 10: Effect of latent dimension (FC-AE)"); plt.tight_layout(); save_show("plot10_latent_dim_study")

# =========================== Consolidated tables ===========================
cons = pd.DataFrame(results).T[["MSE", "MAE", "SSIM", "Parameters", "Time_s"]]
cons.index.name = "Model"
print("\n================ CONSOLIDATED RESULTS ================")
print(cons.round(5).to_string())
print("\n(Denoising CAE row = Gaussian σ=0.2 noisy test input vs clean target; VAE row = reconstruction with z=μ)")

cons.to_csv(f"{OUT}/consolidated_results.csv")
lat_df.to_csv(f"{OUT}/latent_dim_study.csv", index=False)
noise_df.to_csv(f"{OUT}/noise_level_results.csv", index=False)
cmp.to_csv(f"{OUT}/fc_vs_cae.csv", index=False)
vae_table.to_csv(f"{OUT}/vae_metrics.csv", index=False)
print(f"\nAll figures and CSV tables saved in: {OUT}")
```
