```python
# CS3807 - Deep Learning Laboratory - Experiment 4
# Comparative Study of Deep CNN Architectures Using Transfer Learning (CIFAR-10)
# LeNet-5 / AlexNet: trained from scratch (no official ImageNet-pretrained weights exist for either)
# VGG16 / GoogleNet(InceptionV3) / ResNet50: real transfer learning (frozen -> fine-tuned)

import os, time, random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.applications import VGG16, ResNet50, InceptionV3, MobileNetV2
from tensorflow.keras.applications.vgg16 import preprocess_input as vgg16_preprocess
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet50_preprocess
from tensorflow.keras.applications.inception_v3 import preprocess_input as inception_preprocess
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as mobilenet_preprocess
from tensorflow.keras.optimizers import Adam, SGD
from sklearn.metrics import classification_report, confusion_matrix, precision_score, recall_score, f1_score

SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

PLOT_DIR = "/kaggle/working/plots"
os.makedirs(PLOT_DIR, exist_ok=True)
print("GPUs visible:", tf.config.list_physical_devices("GPU"))

NUM_CLASSES = 10
BATCH_SIZE = 32
EPOCHS_SCRATCH = 10      # LeNet-5 / AlexNet (from scratch)
EPOCHS_TL = 5            # frozen-backbone phase for VGG16/ResNet50/InceptionV3
FT_EPOCHS_TL = 4         # fine-tuning phase
AUTOTUNE = tf.data.AUTOTUNE

# ---------------- Task 1: Dataset Preparation ----------------
(X_train_raw, y_train_raw), (X_test_raw, y_test_raw) = cifar10.load_data()
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

y_train_flat = y_train_raw.flatten()
y_test_flat = y_test_raw.flatten()
y_train_oh = to_categorical(y_train_flat, NUM_CLASSES)
y_test_oh = to_categorical(y_test_flat, NUM_CLASSES)

X_train_norm = X_train_raw.astype("float32") / 255.0
X_test_norm = X_test_raw.astype("float32") / 255.0

print("Train shape:", X_train_raw.shape, "Test shape:", X_test_raw.shape)

fig, axes = plt.subplots(2, 5, figsize=(12, 6))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train_raw[i]); ax.set_title(class_names[y_train_flat[i]]); ax.axis("off")
plt.suptitle("Sample CIFAR-10 Images")
plt.tight_layout()
plt.savefig(f"{PLOT_DIR}/plot0_samples.png", dpi=200, bbox_inches="tight")
plt.show()

# ---------------- Shared helpers ----------------
def make_dataset(X_raw, y_oh, preprocess_fn, target_size, batch_size=BATCH_SIZE, shuffle=False):
    ds = tf.data.Dataset.from_tensor_slices((X_raw, y_oh))
    if shuffle:
        ds = ds.shuffle(10000, seed=SEED)
    def _prep(img, label):
        img = tf.cast(img, tf.float32)
        if target_size != (32, 32):
            img = tf.image.resize(img, target_size)
        img = preprocess_fn(img)
        return img, label
    return ds.map(_prep, num_parallel_calls=AUTOTUNE).batch(batch_size).prefetch(AUTOTUNE)

def evaluate_model(model, X_raw, y_flat, preprocess_fn, target_size, batch_size=BATCH_SIZE):
    ds = make_dataset(X_raw, to_categorical(y_flat, NUM_CLASSES), preprocess_fn, target_size, batch_size)
    probs = model.predict(ds, verbose=0)
    preds = np.argmax(probs, axis=1)
    acc = (preds == y_flat).mean()
    return acc, preds

# ---------------- LeNet-5 (from scratch) ----------------
def build_lenet5():
    return models.Sequential([
        layers.Input(shape=(32, 32, 3)),
        layers.Conv2D(6, (5, 5), activation="tanh", padding="same"),
        layers.AveragePooling2D((2, 2)),
        layers.Conv2D(16, (5, 5), activation="tanh"),
        layers.AveragePooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(120, activation="tanh"),
        layers.Dense(84, activation="tanh"),
        layers.Dense(NUM_CLASSES, activation="softmax"),
    ], name="LeNet5")

lenet = build_lenet5()
lenet.compile(optimizer=Adam(learning_rate=0.001), loss="categorical_crossentropy", metrics=["accuracy"])
t0 = time.time()
lenet_hist = lenet.fit(X_train_norm, y_train_oh, batch_size=BATCH_SIZE, epochs=EPOCHS_SCRATCH,
                        validation_data=(X_test_norm, y_test_oh), verbose=1)
lenet_time = time.time() - t0
lenet_params = lenet.count_params()
lenet_test_acc = max(lenet_hist.history["val_accuracy"])
print("LeNet-5 params:", lenet_params, "| best val acc:", lenet_test_acc, "| time:", lenet_time)

# ---------------- AlexNet-style (scaled down for 32x32, from scratch) ----------------
def build_alexnet_mini():
    return models.Sequential([
        layers.Input(shape=(32, 32, 3)),
        layers.Conv2D(64, (3, 3), strides=1, activation="relu", padding="same"),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(192, (3, 3), activation="relu", padding="same"),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(384, (3, 3), activation="relu", padding="same"),
        layers.Conv2D(256, (3, 3), activation="relu", padding="same"),
        layers.Conv2D(256, (3, 3), activation="relu", padding="same"),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(1024, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(1024, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(NUM_CLASSES, activation="softmax"),
    ], name="AlexNetMini")

alexnet = build_alexnet_mini()
alexnet.compile(optimizer=Adam(learning_rate=0.001), loss="categorical_crossentropy", metrics=["accuracy"])
t0 = time.time()
alexnet_hist = alexnet.fit(X_train_norm, y_train_oh, batch_size=BATCH_SIZE, epochs=EPOCHS_SCRATCH,
                            validation_data=(X_test_norm, y_test_oh), verbose=1)
alexnet_time = time.time() - t0
alexnet_params = alexnet.count_params()
alexnet_test_acc = max(alexnet_hist.history["val_accuracy"])
print("AlexNet-mini params:", alexnet_params, "| best val acc:", alexnet_test_acc, "| time:", alexnet_time)

# ---------------- Generic transfer-learning builder (Task 2) ----------------
def build_transfer_model(base_fn, input_shape, dense_units=128, name="TL"):
    base = base_fn(weights="imagenet", include_top=False, input_shape=input_shape)
    base.trainable = False
    inp = layers.Input(shape=input_shape)
    h = base(inp, training=False)
    h = layers.GlobalAveragePooling2D()(h)
    h = layers.Dense(dense_units, activation="relu")(h)
    out = layers.Dense(NUM_CLASSES, activation="softmax")(h)
    return models.Model(inp, out, name=name), base

def train_transfer(model, base, train_ds, val_ds, unfreeze_from_frac=0.8):
    model.compile(optimizer=Adam(learning_rate=0.001), loss="categorical_crossentropy", metrics=["accuracy"])
    t0 = time.time()
    hist_frozen = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS_TL, verbose=1)
    frozen_time = time.time() - t0
    frozen_best_acc = max(hist_frozen.history["val_accuracy"])

    n_layers = len(base.layers)
    cutoff = int(n_layers * unfreeze_from_frac)
    for layer in base.layers[:cutoff]:
        layer.trainable = False
    for layer in base.layers[cutoff:]:
        layer.trainable = True
    model.compile(optimizer=Adam(learning_rate=1e-4), loss="categorical_crossentropy", metrics=["accuracy"])
    t0 = time.time()
    hist_ft = model.fit(train_ds, validation_data=val_ds, epochs=FT_EPOCHS_TL, verbose=1)
    ft_time = time.time() - t0
    ft_best_acc = max(hist_ft.history["val_accuracy"])

    return dict(hist_frozen=hist_frozen.history, hist_ft=hist_ft.history,
                frozen_best_acc=frozen_best_acc, ft_best_acc=ft_best_acc,
                total_time=frozen_time + ft_time, params=model.count_params())

# ---------------- VGG16 (transfer learning) ----------------
vgg_train_ds = make_dataset(X_train_raw, y_train_oh, vgg16_preprocess, (32, 32), shuffle=True)
vgg_val_ds = make_dataset(X_test_raw, y_test_oh, vgg16_preprocess, (32, 32))
vgg_model, vgg_base = build_transfer_model(VGG16, (32, 32, 3), name="VGG16_TL")
vgg_results = train_transfer(vgg_model, vgg_base, vgg_train_ds, vgg_val_ds)
print("VGG16 -> frozen best:", vgg_results["frozen_best_acc"], "| fine-tuned best:", vgg_results["ft_best_acc"])

# ---------------- ResNet50 (transfer learning) ----------------
resnet_train_ds = make_dataset(X_train_raw, y_train_oh, resnet50_preprocess, (32, 32), shuffle=True)
resnet_val_ds = make_dataset(X_test_raw, y_test_oh, resnet50_preprocess, (32, 32))
resnet_model, resnet_base = build_transfer_model(ResNet50, (32, 32, 3), name="ResNet50_TL")
resnet_results = train_transfer(resnet_model, resnet_base, resnet_train_ds, resnet_val_ds)
print("ResNet50 -> frozen best:", resnet_results["frozen_best_acc"], "| fine-tuned best:", resnet_results["ft_best_acc"])

# ---------------- GoogleNet / InceptionV3 (transfer learning; needs >=75x75 input) ----------------
inception_train_ds = make_dataset(X_train_raw, y_train_oh, inception_preprocess, (75, 75), shuffle=True)
inception_val_ds = make_dataset(X_test_raw, y_test_oh, inception_preprocess, (75, 75))
inception_model, inception_base = build_transfer_model(InceptionV3, (75, 75, 3), name="InceptionV3_TL")
inception_results = train_transfer(inception_model, inception_base, inception_train_ds, inception_val_ds)
print("GoogleNet(InceptionV3) -> frozen best:", inception_results["frozen_best_acc"],
      "| fine-tuned best:", inception_results["ft_best_acc"])

# ---------------- Mandatory Plots: Accuracy/Loss curves for each transfer-learning model ----------------
def plot_curves(results, title, fname):
    hf, hft = results["hist_frozen"], results["hist_ft"]
    acc = hf["accuracy"] + hft["accuracy"]; val_acc = hf["val_accuracy"] + hft["val_accuracy"]
    loss = hf["loss"] + hft["loss"]; val_loss = hf["val_loss"] + hft["val_loss"]

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
    ax1.plot(acc, label="Train Accuracy"); ax1.plot(val_acc, "--", label="Validation Accuracy")
    ax1.axvline(len(hf["accuracy"]) - 0.5, color="gray", linestyle=":", label="Fine-tune start")
    ax1.set_title(f"{title}: Accuracy"); ax1.set_xlabel("Epoch"); ax1.legend(); ax1.grid(True)

    ax2.plot(loss, label="Train Loss"); ax2.plot(val_loss, "--", label="Validation Loss")
    ax2.axvline(len(hf["loss"]) - 0.5, color="gray", linestyle=":", label="Fine-tune start")
    ax2.set_title(f"{title}: Loss"); ax2.set_xlabel("Epoch"); ax2.legend(); ax2.grid(True)

    plt.tight_layout()
    plt.savefig(f"{PLOT_DIR}/{fname}.png", dpi=200, bbox_inches="tight")
    plt.show()

plot_curves(vgg_results, "VGG16", "plot1_vgg16_curves")
plot_curves(resnet_results, "ResNet50", "plot2_resnet50_curves")
plot_curves(inception_results, "GoogleNet (InceptionV3)", "plot3_inception_curves")

# ---------------- Task 5: Evaluation of the best transfer-learning model ----------------
best_name, best_model, best_preprocess, best_size, best_results = max(
    [("VGG16", vgg_model, vgg16_preprocess, (32, 32), vgg_results),
     ("ResNet50", resnet_model, resnet50_preprocess, (32, 32), resnet_results),
     ("GoogleNet(InceptionV3)", inception_model, inception_preprocess, (75, 75), inception_results)],
    key=lambda t: t[4]["ft_best_acc"],
)
print("Best transfer-learning model:", best_name)

test_acc, preds = evaluate_model(best_model, X_test_raw, y_test_flat, best_preprocess, best_size)
precision = precision_score(y_test_flat, preds, average="weighted")
recall = recall_score(y_test_flat, preds, average="weighted")
f1 = f1_score(y_test_flat, preds, average="weighted")

print("================ Classification Report:", best_name, "================")
print(classification_report(y_test_flat, preds, target_names=class_names))

cm = confusion_matrix(y_test_flat, preds)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_names, yticklabels=class_names)
plt.title(f"Confusion Matrix ({best_name})"); plt.xlabel("Predicted"); plt.ylabel("True")
plt.tight_layout()
plt.savefig(f"{PLOT_DIR}/plot4_confusion_matrix.png", dpi=200, bbox_inches="tight")
plt.show()

mis_idx = np.where(preds != y_test_flat)[0][:10]
plt.figure(figsize=(12, 5))
for i, idx in enumerate(mis_idx):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_test_raw[idx])
    plt.title(f"T:{class_names[y_test_flat[idx]]}\nP:{class_names[preds[idx]]}", fontsize=8)
    plt.axis("off")
plt.suptitle(f"Misclassified Examples ({best_name})")
plt.tight_layout()
plt.savefig(f"{PLOT_DIR}/plot5_misclassified.png", dpi=200, bbox_inches="tight")
plt.show()

final_report = pd.Series({
    "Model": best_name,
    "Training Accuracy (last epoch)": best_results["hist_ft"]["accuracy"][-1],
    "Testing Accuracy": test_acc,
    "Precision": precision,
    "Recall": recall,
    "F1-score": f1,
    "Training Time (s)": best_results["total_time"],
    "Total Parameters": best_results["params"],
})
print(final_report)

# ---------------- Section 18.2: Comparison of CNN Architectures ----------------
comparison_table = pd.DataFrame({
    "Model": ["LeNet-5", "AlexNet(mini)", "VGG16", "GoogleNet(InceptionV3)", "ResNet50"],
    "Parameters": [lenet_params, alexnet_params, vgg_results["params"], inception_results["params"], resnet_results["params"]],
    "Accuracy (%)": [lenet_test_acc * 100, alexnet_test_acc * 100,
                      vgg_results["ft_best_acc"] * 100, inception_results["ft_best_acc"] * 100,
                      resnet_results["ft_best_acc"] * 100],
    "Training Time (s)": [lenet_time, alexnet_time, vgg_results["total_time"],
                           inception_results["total_time"], resnet_results["total_time"]],
})
print(comparison_table)

# ---------------- Section 16: Hyperparameter Study (base model: MobileNetV2, fast + 32x32-native) ----------------
def build_mnv2(dense_units=128, freeze_all=True):
    base = MobileNetV2(weights="imagenet", include_top=False, input_shape=(32, 32, 3))
    base.trainable = not freeze_all
    if not freeze_all:
        cutoff = int(len(base.layers) * 0.8)
        for layer in base.layers[:cutoff]:
            layer.trainable = False
    inp = layers.Input(shape=(32, 32, 3))
    h = base(inp, training=False if freeze_all else None)
    h = layers.GlobalAveragePooling2D()(h)
    h = layers.Dense(dense_units, activation="relu")(h)
    out = layers.Dense(NUM_CLASSES, activation="softmax")(h)
    return models.Model(inp, out)

def quick_train(model, optimizer, batch_size, epochs=3):
    train_ds = make_dataset(X_train_raw, y_train_oh, mobilenet_preprocess, (32, 32), batch_size, shuffle=True)
    val_ds = make_dataset(X_test_raw, y_test_oh, mobilenet_preprocess, (32, 32), batch_size)
    model.compile(optimizer=optimizer, loss="categorical_crossentropy", metrics=["accuracy"])
    hist = model.fit(train_ds, validation_data=val_ds, epochs=epochs, verbose=0)
    return max(hist.history["val_accuracy"])

hp_results = {}

for lr in [0.001, 0.0001]:
    m = build_mnv2()
    hp_results[f"lr={lr}"] = quick_train(m, Adam(learning_rate=lr), BATCH_SIZE)
    tf.keras.backend.clear_session()

for bs in [16, 32, 64]:
    m = build_mnv2()
    hp_results[f"batch={bs}"] = quick_train(m, Adam(learning_rate=0.001), bs)
    tf.keras.backend.clear_session()

for opt_name, opt in [("Adam", Adam(learning_rate=0.001)), ("SGD", SGD(learning_rate=0.001, momentum=0.9))]:
    m = build_mnv2()
    hp_results[f"optimizer={opt_name}"] = quick_train(m, opt, BATCH_SIZE)
    tf.keras.backend.clear_session()

for du in [128, 256]:
    m = build_mnv2(dense_units=du)
    hp_results[f"dense_units={du}"] = quick_train(m, Adam(learning_rate=0.001), BATCH_SIZE)
    tf.keras.backend.clear_session()

for frz_name, frz in [("All Frozen", True), ("Partial (fine-tune)", False)]:
    m = build_mnv2(freeze_all=frz)
    hp_results[f"frozen={frz_name}"] = quick_train(m, Adam(learning_rate=0.001), BATCH_SIZE)
    tf.keras.backend.clear_session()

print(pd.Series(hp_results))

plt.figure(figsize=(10, 5))
plt.bar(list(hp_results.keys()), [v * 100 for v in hp_results.values()])
plt.xticks(rotation=45, ha="right")
plt.ylabel("Best Validation Accuracy (%)")
plt.title("Hyperparameter Study (MobileNetV2 backbone)")
plt.tight_layout()
plt.savefig(f"{PLOT_DIR}/plot6_hyperparameter_study.png", dpi=200, bbox_inches="tight")
plt.show()
```
