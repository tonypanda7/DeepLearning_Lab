```python
# CS3807 – Deep Learning Laboratory — Experiment 5
# MobileNetV2 transfer learning study on Oxford-IIIT Pet — single-cell Kaggle version

import os
import time
import random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow_datasets as tfds

from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.optimizers import SGD, RMSprop, Adam
from tensorflow.keras.regularizers import l2
from sklearn.model_selection import KFold
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix

SEED = 7
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

PLOT_DIR = "/kaggle/working/plots"
os.makedirs(PLOT_DIR, exist_ok=True)
print("GPUs visible:", tf.config.list_physical_devices("GPU"))

IMG_SIZE, NUM_CLASSES = 224, 37
EPOCHS, FT_EPOCHS = 5, 5

# ---- 1. Load & prepare dataset ----
raw_train, raw_test = tfds.load("oxford_iiit_pet", split=["train", "test"], as_supervised=True, shuffle_files=True)
full_ds = raw_train.concatenate(raw_test)

def resize_and_cast(img, label):
    return tf.image.resize(img, (IMG_SIZE, IMG_SIZE)), tf.cast(label, tf.int32)

full_ds = full_ds.map(resize_and_cast, num_parallel_calls=tf.data.AUTOTUNE)
full_ds = full_ds.shuffle(6000, seed=SEED, reshuffle_each_iteration=False)

all_images, all_labels = [], []
for img, lbl in full_ds:
    all_images.append(img.numpy()); all_labels.append(lbl.numpy())
all_images = np.asarray(all_images, dtype=np.float32)
all_labels = np.asarray(all_labels, dtype=np.int32)
print("Total images:", all_images.shape, "| labels:", all_labels.shape)

rng = np.random.default_rng(SEED)
perm = rng.permutation(len(all_images))
all_images, all_labels = all_images[perm], all_labels[perm]

n = len(all_images)
train_cut, val_cut = int(0.70 * n), int(0.85 * n)
X_train, y_train = all_images[:train_cut], all_labels[:train_cut]
X_val, y_val     = all_images[train_cut:val_cut], all_labels[train_cut:val_cut]
X_test, y_test   = all_images[val_cut:], all_labels[val_cut:]

X_train, X_val, X_test = preprocess_input(X_train), preprocess_input(X_val), preprocess_input(X_test)
print("Train:", X_train.shape, "Val:", X_val.shape, "Test:", X_test.shape)

def to_tf_dataset(X, y, batch_size=32, shuffle=False):
    ds = tf.data.Dataset.from_tensor_slices((X, y))
    if shuffle:
        ds = ds.shuffle(4096, seed=SEED)
    return ds.batch(batch_size).prefetch(tf.data.AUTOTUNE)

train_ds = to_tf_dataset(X_train, y_train, 32, shuffle=True)
val_ds   = to_tf_dataset(X_val, y_val, 32)
test_ds  = to_tf_dataset(X_test, y_test, 32)

# ---- 2. Model builder ----
def build_classifier(init="he_normal", reg=None, drop=0.0, use_bn=False, unfreeze_base=False):
    backbone = MobileNetV2(weights="imagenet", include_top=False, input_shape=(IMG_SIZE, IMG_SIZE, 3))
    backbone.trainable = unfreeze_base
    inp = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
    h = backbone(inp, training=False)
    h = layers.GlobalAveragePooling2D()(h)
    if use_bn:
        h = layers.BatchNormalization()(h)
    h = layers.Dense(128, activation="relu", kernel_initializer=init, kernel_regularizer=reg)(h)
    if drop > 0:
        h = layers.Dropout(drop)(h)
    out = layers.Dense(NUM_CLASSES, activation="softmax", kernel_initializer=init)(h)
    return Model(inp, out, name="mobilenetv2_pet_classifier")

def compile_and_train(model, optimizer, epochs=EPOCHS, train_data=train_ds, verbose=1):
    model.compile(optimizer=optimizer, loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    t0 = time.time()
    hist = model.fit(train_data, validation_data=val_ds, epochs=epochs, verbose=verbose)
    return hist, time.time() - t0

# ---- 3. Weight initialization study (Plots 1-2) ----
init_schemes = {"Zero": "zeros", "Random": "random_normal", "Xavier": "glorot_uniform", "He": "he_normal"}
init_hist, init_best_acc = {}, {}
for name, scheme in init_schemes.items():
    model = build_classifier(init=scheme)
    hist, _ = compile_and_train(model, Adam(learning_rate=1e-3))
    init_hist[name] = hist.history
    init_best_acc[name] = max(hist.history["val_accuracy"])
    tf.keras.backend.clear_session()
print(init_best_acc)

plt.figure(figsize=(7, 5))
for name, h in init_hist.items(): plt.plot(h["loss"], label=name)
plt.xlabel("Epoch"); plt.ylabel("Training Loss"); plt.title("Plot 1: Training Loss by Initialization")
plt.legend(); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot1_init_train_loss.png", dpi=300, bbox_inches="tight"); plt.show()

plt.figure(figsize=(7, 5))
for name, h in init_hist.items(): plt.plot(np.array(h["val_accuracy"]) * 100, label=name)
plt.xlabel("Epoch"); plt.ylabel("Validation Accuracy (%)"); plt.title("Plot 2: Validation Accuracy by Initialization")
plt.legend(); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot2_init_val_acc.png", dpi=300, bbox_inches="tight"); plt.show()

# ---- 4. Regularization study (Plots 3-4) ----
reg_configs = {"No Regularization": {}, "L2": {"reg": l2(1e-3)}, "Dropout": {"drop": 0.5}, "Batch Normalization": {"use_bn": True}}
reg_hist, reg_best_acc = {}, {}
for name, cfg in reg_configs.items():
    model = build_classifier(**cfg)
    hist, _ = compile_and_train(model, Adam(learning_rate=1e-3))
    reg_hist[name] = hist.history
    reg_best_acc[name] = max(hist.history["val_accuracy"])
    tf.keras.backend.clear_session()
print(reg_best_acc)

plt.figure(figsize=(7, 5))
for name, h in reg_hist.items():
    plt.plot(np.array(h["accuracy"]) * 100, "--", label=f"{name} (train)")
    plt.plot(np.array(h["val_accuracy"]) * 100, label=f"{name} (val)")
plt.xlabel("Epoch"); plt.ylabel("Accuracy (%)"); plt.title("Plot 3: Train vs Val Accuracy by Regularization")
plt.legend(fontsize=8); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot3_reg_accuracy.png", dpi=300, bbox_inches="tight"); plt.show()

plt.figure(figsize=(7, 5))
for name, h in reg_hist.items():
    plt.plot(h["loss"], "--", label=f"{name} (train)")
    plt.plot(h["val_loss"], label=f"{name} (val)")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Plot 4: Train vs Val Loss by Regularization")
plt.legend(fontsize=8); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot4_reg_loss.png", dpi=300, bbox_inches="tight"); plt.show()

# ---- 5. Batch norm with vs without (Plot 5) ----
plt.figure(figsize=(7, 5))
plt.plot(np.array(reg_hist["No Regularization"]["val_accuracy"]) * 100, label="Without BN")
plt.plot(np.array(reg_hist["Batch Normalization"]["val_accuracy"]) * 100, label="With BN")
plt.xlabel("Epoch"); plt.ylabel("Validation Accuracy (%)"); plt.title("Plot 5: With vs Without Batch Normalization")
plt.legend(); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot5_batchnorm.png", dpi=300, bbox_inches="tight"); plt.show()

# ---- 6. Optimizer comparison (Plots 6-7) ----
optimizer_bank = {"SGD": lambda: SGD(learning_rate=1e-3), "Momentum": lambda: SGD(learning_rate=1e-3, momentum=0.9),
                   "RMSProp": lambda: RMSprop(learning_rate=1e-3), "Adam": lambda: Adam(learning_rate=1e-3)}
opt_hist, opt_summary = {}, {}
for name, make_opt in optimizer_bank.items():
    model = build_classifier()
    hist, elapsed = compile_and_train(model, make_opt())
    opt_hist[name] = hist.history
    opt_summary[name] = {"Final Loss": hist.history["loss"][-1], "Best Val Accuracy": max(hist.history["val_accuracy"]), "Time (s)": elapsed}
    tf.keras.backend.clear_session()
optimizer_table = pd.DataFrame(opt_summary).T
print(optimizer_table)

plt.figure(figsize=(7, 5))
for name, h in opt_hist.items(): plt.plot(h["loss"], label=name)
plt.xlabel("Epoch"); plt.ylabel("Training Loss"); plt.title("Plot 6: Training Loss by Optimizer")
plt.legend(); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot6_optimizer_loss.png", dpi=300, bbox_inches="tight"); plt.show()

plt.figure(figsize=(7, 5))
for name, h in opt_hist.items(): plt.plot(np.array(h["val_accuracy"]) * 100, label=name)
plt.xlabel("Epoch"); plt.ylabel("Validation Accuracy (%)"); plt.title("Plot 7: Validation Accuracy by Optimizer")
plt.legend(); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot7_optimizer_val_acc.png", dpi=300, bbox_inches="tight"); plt.show()

# ---- 7. Hyperparameter tuning (Plots 8-10) ----
lr_results = {}
for lr in [1e-3, 1e-4]:
    model = build_classifier()
    hist, _ = compile_and_train(model, Adam(learning_rate=lr))
    lr_results[lr] = max(hist.history["val_accuracy"])
    tf.keras.backend.clear_session()
print(lr_results)
plt.figure(figsize=(6, 4))
plt.plot(list(map(str, lr_results.keys())), [v * 100 for v in lr_results.values()], marker="o")
plt.xlabel("Learning Rate"); plt.ylabel("Best Validation Accuracy (%)"); plt.title("Plot 8: LR vs Validation Accuracy")
plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot8_lr.png", dpi=300, bbox_inches="tight"); plt.show()

batch_results = {}
for bs in [16, 32, 64]:
    ds = to_tf_dataset(X_train, y_train, bs, shuffle=True)
    model = build_classifier()
    hist, _ = compile_and_train(model, Adam(learning_rate=1e-3), train_data=ds)
    batch_results[bs] = max(hist.history["val_accuracy"])
    tf.keras.backend.clear_session()
print(batch_results)
plt.figure(figsize=(6, 4))
plt.plot(list(map(str, batch_results.keys())), [v * 100 for v in batch_results.values()], marker="o")
plt.xlabel("Batch Size"); plt.ylabel("Best Validation Accuracy (%)"); plt.title("Plot 9: Batch Size vs Validation Accuracy")
plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot9_batch_size.png", dpi=300, bbox_inches="tight"); plt.show()

dropout_results = {}
for d in [0.0, 0.25, 0.5]:
    model = build_classifier(drop=d)
    hist, _ = compile_and_train(model, Adam(learning_rate=1e-3))
    dropout_results[d] = max(hist.history["val_accuracy"])
    tf.keras.backend.clear_session()
print(dropout_results)
plt.figure(figsize=(6, 4))
plt.plot(list(map(str, dropout_results.keys())), [v * 100 for v in dropout_results.values()], marker="o")
plt.xlabel("Dropout Rate"); plt.ylabel("Best Validation Accuracy (%)"); plt.title("Plot 10: Dropout vs Validation Accuracy")
plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot10_dropout.png", dpi=300, bbox_inches="tight"); plt.show()

# ---- 8. Feature extraction vs fine-tuning (Plots 11-12) ----
tl_model = build_classifier(unfreeze_base=False)
fe_hist, fe_time = compile_and_train(tl_model, Adam(learning_rate=1e-3))

backbone_layers = tl_model.layers[1].layers
for layer in backbone_layers[:-30]: layer.trainable = False
for layer in backbone_layers[-30:]: layer.trainable = True
tl_model.compile(optimizer=Adam(learning_rate=1e-4), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
t0 = time.time()
ft_hist = tl_model.fit(train_ds, validation_data=val_ds, epochs=FT_EPOCHS, verbose=1)
ft_time = time.time() - t0

plt.figure(figsize=(7, 5))
plt.plot(np.array(fe_hist.history["val_accuracy"]) * 100, label="Feature Extraction")
plt.plot(np.array(ft_hist.history["val_accuracy"]) * 100, label="Fine-Tuning")
plt.xlabel("Epoch"); plt.ylabel("Validation Accuracy (%)"); plt.title("Plot 11: Feature Extraction vs Fine-Tuning")
plt.legend(); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot11_transfer_learning.png", dpi=300, bbox_inches="tight"); plt.show()

plt.figure(figsize=(7, 5))
plt.plot(fe_hist.history["loss"], label="Feature Extraction (train)")
plt.plot(fe_hist.history["val_loss"], label="Feature Extraction (val)")
plt.plot(ft_hist.history["loss"], label="Fine-Tuning (train)")
plt.plot(ft_hist.history["val_loss"], label="Fine-Tuning (val)")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Plot 12: Loss Before/After Fine-Tuning")
plt.legend(fontsize=8); plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot12_finetune_loss.png", dpi=300, bbox_inches="tight"); plt.show()

# ---- 9. 5-fold cross-validation (Plot 13) ----
candidate_configs = {"C1": dict(lr=1e-3, drop=0.0, batch=32), "C2": dict(lr=1e-3, drop=0.25, batch=32),
                      "C3": dict(lr=1e-4, drop=0.25, batch=32), "C4": dict(lr=1e-4, drop=0.5, batch=32)}
print(candidate_configs)

def make_fold_dataset(indices, batch_size):
    def gen():
        idx = indices.copy()
        rng2 = np.random.default_rng(SEED)
        rng2.shuffle(idx)
        for start in range(0, len(idx), batch_size):
            b = idx[start:start + batch_size]
            yield X_train[b], y_train[b]
    sig = (tf.TensorSpec(shape=(None, IMG_SIZE, IMG_SIZE, 3), dtype=tf.float32), tf.TensorSpec(shape=(None,), dtype=tf.int32))
    return tf.data.Dataset.from_generator(gen, output_signature=sig)

cv_scores = {}
kfold = KFold(n_splits=5, shuffle=True, random_state=SEED)
for cname, cfg in candidate_configs.items():
    fold_acc = []
    for tr_idx, va_idx in kfold.split(X_train):
        fold_train = make_fold_dataset(tr_idx, cfg["batch"])
        fold_val = make_fold_dataset(va_idx, cfg["batch"])
        m = build_classifier(drop=cfg["drop"])
        m.compile(optimizer=Adam(learning_rate=cfg["lr"]), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
        h = m.fit(fold_train, validation_data=fold_val, epochs=3,
                  steps_per_epoch=int(np.ceil(len(tr_idx) / cfg["batch"])),
                  validation_steps=int(np.ceil(len(va_idx) / cfg["batch"])), verbose=0)
        fold_acc.append(max(h.history["val_accuracy"]))
        del m, fold_train, fold_val, h
        tf.keras.backend.clear_session()
    cv_scores[cname] = fold_acc + [float(np.mean(fold_acc)), float(np.std(fold_acc))]

cv_table = pd.DataFrame(cv_scores, index=["F1", "F2", "F3", "F4", "F5", "Mean", "SD"]).T
print(cv_table)

plt.figure(figsize=(7, 5))
plt.bar(cv_table.index, cv_table["Mean"] * 100, yerr=cv_table["SD"] * 100, capsize=5)
plt.xlabel("Configuration"); plt.ylabel("Mean Validation Accuracy (%)"); plt.title("Plot 13: 5-Fold CV Accuracy by Configuration")
plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot13_cv_accuracy.png", dpi=300, bbox_inches="tight"); plt.show()

best_config_name = cv_table["Mean"].idxmax()
print("Selected configuration:", best_config_name, dict(candidate_configs[best_config_name]))

# ---- 10. Final model: retrain + test evaluation (Plot 14) ----
best_cfg = candidate_configs[best_config_name]
final_model = build_classifier(drop=best_cfg["drop"])
final_model.compile(optimizer=Adam(learning_rate=best_cfg["lr"]), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
t0 = time.time()
final_hist = final_model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS, verbose=1)
final_train_time = time.time() - t0

test_loss, test_acc = final_model.evaluate(test_ds, verbose=0)
probs = final_model.predict(test_ds)
preds = np.argmax(probs, axis=1)
precision = precision_score(y_test, preds, average="weighted")
recall = recall_score(y_test, preds, average="weighted")
f1 = f1_score(y_test, preds, average="weighted")

report = pd.Series({"Mean CV Accuracy": cv_table.loc[best_config_name, "Mean"], "CV Standard Deviation": cv_table.loc[best_config_name, "SD"],
                     "Test Accuracy": test_acc, "Precision": precision, "Recall": recall, "F1-score": f1,
                     "Training Time (s)": final_train_time, "Number of Parameters": final_model.count_params()})
print(report)

cm = confusion_matrix(y_test, preds)
plt.figure(figsize=(11, 9))
sns.heatmap(cm, annot=False, cmap="viridis", xticklabels=range(NUM_CLASSES), yticklabels=range(NUM_CLASSES))
plt.xlabel("Predicted Label"); plt.ylabel("True Label"); plt.title("Plot 14: Confusion Matrix")
plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot14_confusion_matrix.png", dpi=300, bbox_inches="tight"); plt.show()

# ---- 11. Optional: misclassified examples (Plot 15) ----
mis_idx = np.where(preds != y_test)[0][:10]
plt.figure(figsize=(12, 5))
for i, idx in enumerate(mis_idx):
    plt.subplot(2, 5, i + 1)
    plt.imshow(np.clip((X_test[idx] + 1) / 2, 0, 1))
    plt.title(f"T:{y_test[idx]} P:{preds[idx]}", fontsize=9); plt.axis("off")
plt.tight_layout(); plt.savefig(f"{PLOT_DIR}/plot15_misclassified.png", dpi=300, bbox_inches="tight"); plt.show()
```
